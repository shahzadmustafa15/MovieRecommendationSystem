# Movie Recommendation System
## Phase 3: Data Preprocessing and Feature Engineering

The objective of this notebook is to prepare the MovieLens 1M
dataset for recommendation algorithm development.

We will validate the raw data, handle data quality issues,
transform movie genres into numerical features, and prepare
user-movie rating data for collaborative filtering.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
DATA_DIR = Path("../data")

movies = pd.read_csv(
    DATA_DIR / "movies.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["MovieID", "Title", "Genres"]
)

ratings = pd.read_csv(
    DATA_DIR / "ratings.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["UserID", "MovieID", "Rating", "Timestamp"]
)

users = pd.read_csv(
    DATA_DIR / "users.dat",
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["UserID", "Gender", "Age", "Occupation", "ZipCode"]
)

print("Datasets loaded successfully!")
```

```python
movies_clean = movies.copy()
ratings_clean = ratings.copy()
users_clean = users.copy()

print("Working copies created.")
```

```python
print("Movies missing values:")
print(movies_clean.isnull().sum())

print("\nRatings missing values:")
print(ratings_clean.isnull().sum())

print("\nUsers missing values:")
print(users_clean.isnull().sum())
```

```python
print("Movies duplicates:", movies_clean.duplicated().sum())
print("Ratings duplicates:", ratings_clean.duplicated().sum())
print("Users duplicates:", users_clean.duplicated().sum())

print(
    "Repeated user-movie combinations:",
    ratings_clean.duplicated(
        subset=["UserID", "MovieID"]
    ).sum()
)
```

```python
print("Movies duplicates:", movies_clean.duplicated().sum())
print("Ratings duplicates:", ratings_clean.duplicated().sum())
print("Users duplicates:", users_clean.duplicated().sum())

print(
    "Repeated user-movie combinations:",
    ratings_clean.duplicated(
        subset=["UserID", "MovieID"]
    ).sum()
)
```

```python
movies_clean["MovieID"] = movies_clean["MovieID"].astype("int32")
movies_clean["Title"] = movies_clean["Title"].astype("string")
movies_clean["Genres"] = movies_clean["Genres"].astype("string")

movies_clean["Title"] = movies_clean["Title"].str.strip()
movies_clean["Genres"] = movies_clean["Genres"].str.strip()

movies_clean.info()
```

```python
ratings_clean["UserID"] = ratings_clean["UserID"].astype("int32")
ratings_clean["MovieID"] = ratings_clean["MovieID"].astype("int32")
ratings_clean["Rating"] = ratings_clean["Rating"].astype("float32")

ratings_clean["Timestamp"] = pd.to_numeric(
    ratings_clean["Timestamp"],
    errors="raise"
)

ratings_clean.info()
```

```python
users_clean["UserID"] = users_clean["UserID"].astype("int32")
users_clean["Gender"] = users_clean["Gender"].astype("string")
users_clean["Age"] = users_clean["Age"].astype("int32")
users_clean["Occupation"] = users_clean["Occupation"].astype("int32")
users_clean["ZipCode"] = users_clean["ZipCode"].astype("string")

users_clean.info()
```

```python
ratings_clean["RatingDate"] = pd.to_datetime(
    ratings_clean["Timestamp"],
    unit="s"
)

ratings_clean[
    ["UserID", "MovieID", "Rating", "Timestamp", "RatingDate"]
].head()
```

```python
invalid_movies = ~ratings_clean["MovieID"].isin(
    movies_clean["MovieID"]
)

invalid_users = ~ratings_clean["UserID"].isin(
    users_clean["UserID"]
)

print("Invalid movie references:", invalid_movies.sum())
print("Invalid user references:", invalid_users.sum())
```

```python
PROCESSED_DIR = DATA_DIR / "processed"
PROCESSED_DIR.mkdir(exist_ok=True)

movies_clean.to_csv(
    PROCESSED_DIR / "movies_clean.csv",
    index=False
)

ratings_clean.to_csv(
    PROCESSED_DIR / "ratings_clean.csv",
    index=False
)

users_clean.to_csv(
    PROCESSED_DIR / "users_clean.csv",
    index=False
)

print("Processed datasets saved successfully!")
```
