# Movie Recommendation System
## Phase 1: Data Understanding

This notebook is the first step of our Movie Recommendation System project.

The objective of this notebook is to load, inspect, and understand the MovieLens 1M dataset before performing data cleaning, exploratory data analysis, and model development.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
from pathlib import Path

DATA_DIR = Path("../data")

movies_path = DATA_DIR / "movies.dat"
ratings_path = DATA_DIR / "ratings.dat"
users_path = DATA_DIR / "users.dat"

print("Dataset paths defined.")
```

```python
for file_path in [movies_path, ratings_path, users_path]:
    print(f"{file_path.name}: {'Found' if file_path.exists() else 'Missing'}")
```

```python
movies = pd.read_csv(
    movies_path,
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["MovieID", "Title", "Genres"]
)

movies.head()
```

```python
ratings = pd.read_csv(
    ratings_path,
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["UserID", "MovieID", "Rating", "Timestamp"]
)

ratings.head()
```

```python
users = pd.read_csv(
    users_path,
    sep="::",
    engine="python",
    encoding="latin-1",
    names=["UserID", "Gender", "Age", "Occupation", "ZipCode"]
)

users.head()
```

```python
print("Movies dataset:", movies.shape)
print("Ratings dataset:", ratings.shape)
print("Users dataset:", users.shape)
```

```python
print("Movies columns:")
print(movies.columns.tolist())

print("\nRatings columns:")
print(ratings.columns.tolist())

print("\nUsers columns:")
print(users.columns.tolist())
```

```python
movies.isnull().sum()
```

```python
ratings.isnull().sum()
```

```python
users.isnull().sum()
```

```python
print("Duplicate movie records:", movies.duplicated().sum())
```

```python
print("Duplicate rating records:", ratings.duplicated().sum())
```

```python
print("Duplicate user records:", users.duplicated().sum())
```

```python
duplicate_ratings = ratings.duplicated(
    subset=["UserID", "MovieID"]
).sum()

print("Repeated user-movie combinations:", duplicate_ratings)
```

```python
print("Unique users:", ratings["UserID"].nunique())
print("Unique movies rated:", ratings["MovieID"].nunique())
print("Total rating records:", len(ratings))
```

```python
print("Minimum rating:", ratings["Rating"].min())
print("Maximum rating:", ratings["Rating"].max())

print("\nAvailable rating values:")
print(sorted(ratings["Rating"].unique()))
```

```python
invalid_movie_ids = ~ratings["MovieID"].isin(movies["MovieID"])

print("Ratings with invalid movie IDs:", invalid_movie_ids.sum())
```

```python
invalid_user_ids = ~ratings["UserID"].isin(users["UserID"])

print("Ratings with invalid user IDs:", invalid_user_ids.sum())
```

- Missing values

- Repeated user-movie combinations
